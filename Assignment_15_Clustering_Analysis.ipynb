```python
### dropping ID from the dataframe
df = df.drop(columns=['ID#'])
```

```python
X = df.drop(columns=['Award?'])
```

```python
### handling skewness 
import numpy as np

skewed_cols = [
    'Balance',
    'Qual_miles',
    'Bonus_miles',
    'Flight_miles_12mo',
    'Flight_trans_12'
]

for col in skewed_cols:
    X[col] = np.log1p(X[col])
```

```python
### Scaling data using Standard scaler 
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)

X_scaled = pd.DataFrame(
    X_scaled,
    columns=X.columns
)
```

```python
X_scaled.describe()
```

### Observation

- The standardized data has a mean approximately equal to **0** for all variables.
- The standard deviation of all variables is approximately **1**, confirming successful standardization.
- All features are now on a **common scale**, making them suitable for distance-based algorithms such as K-Means.
- `Balance` still has extreme standardized values, indicating the presence of **outliers**.
- Standardization reduces the effect of different measurement scales across variables.
- The standardized dataset is now ready for **clustering analysis**.

```python
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt

inertia = []

K = range(2, 11)

for k in K:
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
    kmeans.fit(X_scaled)
    inertia.append(kmeans.inertia_)

plt.figure(figsize=(8, 5))
plt.plot(K, inertia, marker='o')
plt.xlabel('Number of Clusters (K)')
plt.ylabel('Within-Cluster Sum of Squares (Inertia)')
plt.title('Elbow Curve for K-Means')
plt.show()
```

```python
## aplication of k means since the elbow point is 5
kmeans = KMeans(n_clusters=5, random_state=42, n_init=10)

kmeans_labels = kmeans.fit_predict(X_scaled)

df['KMeans_Cluster'] = kmeans_labels

print(df['KMeans_Cluster'].value_counts())
```

```python
from sklearn.metrics import silhouette_score

kmeans_silhouette = silhouette_score(X_scaled, kmeans_labels)

print("K-Means Silhouette Score:", kmeans_silhouette)
```

```python
### DBScan 
from sklearn.cluster import DBSCAN
from sklearn.metrics import silhouette_score

eps_values = [0.3, 0.5, 0.7, 1.0]
min_samples_values = [5, 10, 15]

for eps in eps_values:
    for min_samples in min_samples_values:

        dbscan = DBSCAN(
            eps=eps,
            min_samples=min_samples
        )

        labels = dbscan.fit_predict(X_scaled)

        n_clusters = len(set(labels)) - (1 if -1 in labels else 0)

        if n_clusters > 1:
            score = silhouette_score(X_scaled, labels)
            print(
                f"eps={eps}, min_samples={min_samples}, "
                f"clusters={n_clusters}, silhouette={score:.3f}"
            )
        else:
            print(
                f"eps={eps}, min_samples={min_samples}, "
                f"clusters={n_clusters}, silhouette=N/A"
            )
```

### Observation

- DBSCAN achieved its best silhouette score of **0.212** with `eps = 1.0` and `min_samples = 15`.
- Lower `eps` values produced many small clusters and **negative silhouette scores**, indicating poor cluster separation.
- Increasing `eps` improved the silhouette score, with the best results occurring at `eps = 1.0`.
- However, the best DBSCAN score (**0.212**) is lower than the K-Means score (**0.253**).
- Therefore, **K-Means with K = 5 performs better than DBSCAN** for this dataset based on the silhouette score.

```python
## cluster analysis
from sklearn.decomposition import PCA
import matplotlib.pyplot as plt

pca = PCA(n_components=2)

X_pca = pca.fit_transform(X_scaled)

plt.figure(figsize=(8, 6))

plt.scatter(
    X_pca[:, 0],
    X_pca[:, 1],
    c=kmeans_labels,
    cmap='viridis',
    s=20
)

plt.xlabel('Principal Component 1')
plt.ylabel('Principal Component 2')
plt.title('K-Means Clustering Visualization')
plt.colorbar(label='Cluster')
plt.show()
```

```python

pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)


dbscan = DBSCAN(eps=1.0, min_samples=15)
dbscan_labels = dbscan.fit_predict(X_scaled)

plt.figure(figsize=(8, 6))

plt.scatter(
    X_pca[:, 0],
    X_pca[:, 1],
    c=dbscan_labels,
    cmap='viridis',
    s=20
)

plt.xlabel('Principal Component 1')
plt.ylabel('Principal Component 2')
plt.title('DBSCAN Clustering Visualization')
plt.colorbar(label='Cluster')
plt.show()
```

```python
from sklearn.metrics import silhouette_score

kmeans_score = silhouette_score(
    X_scaled,
    kmeans_labels
)

print("K-Means Silhouette Score:", kmeans_score)
```

```python
mask = dbscan_labels != -1

if len(set(dbscan_labels[mask])) > 1:
    dbscan_score = silhouette_score(
        X_scaled[mask],
        dbscan_labels[mask]
    )
    print("DBSCAN Silhouette Score:", dbscan_score)
else:
    dbscan_score = None
    print("DBSCAN does not have enough clusters for Silhouette Score.")
```

```python
print("K-Means Silhouette Score:", kmeans_score)
print("DBSCAN Silhouette Score:", dbscan_score)
```

### Evaluation and Performance Observation

- K-Means achieved a Silhouette Score of **0.2530**.
- DBSCAN achieved a higher Silhouette Score of **0.2773**.
- Since a higher Silhouette Score indicates better-defined and better-separated clusters, **DBSCAN performed slightly better than K-Means**.
- The DBSCAN score indicates relatively better cluster separation, although the overall clustering structure is still moderate.
- K-Means also produced reasonably meaningful clusters, but its lower score suggests greater overlap between clusters.
- Overall, based on the Silhouette Score, **DBSCAN is the better-performing algorithm for this dataset**.

### Exercice 1: Prime Numbers  
#### Write a Python program that checks whether a given number is prime or not. A prime number is a natural number greater than 1 that has no positive divisors other than 1 and itself.

```python
# asking the user for a number to check
a = int(input("Enter any Number"))
# if the nummber is less than 1 then obviously not prime
if a <=1 :
    print(a , " is not a prime number")
elif a >=2 :
    for i in range(2,a): 
        if (a % i) == 0:
            print (a ," is not a prime number")
            break
# checking remainder from 2 to a-1 
    else:
        print(a , " is a prime number")
else:
    print(a ," is not a prime number")
```

### Exercise 2: Product of Random Numbers
#### Develop a Python program that generates two random numbers and asks the user to enter the product of these numbers. The program should then check if the user's answer is correct and display an appropriate message.

```python
# for integer type numbers
# First lets generate 2 random numbers 
import random 
print("enter the desirable range you want the random numbers to be from")
x = int(input("enter range start"))
y = int(input("enter range end"))
a = random.randint(x,y)
b = random.randint(x,y)
# now display that to the user and ask for the calcuated product
print("Your Random numbers are",a,b)
c = int(input("Enter the product that you calculated"))
# check and share whther the results are correct or not 
if (a*b) == c:
    print(c," is the correct product of multiplication")
else:
    print(c, " is incorrect product of multiplication")
    print(a*b, " is the correct product")
```

```python
# for float type numbers
# First generate 2 random numbers
import random 
print("enter the desirable range you want the random numbers to be from")
x = float(input("enter range start"))
y = float(input("enter range end"))
a = random.uniform(x,y)
b = random.uniform(x,y)
# now display that to the user and ask for the calcuated product
print("Your Random numbers are",a,b)
c = float(input("Enter the product that you calculated"))
# check and share whther the results are correct or not 
if (a*b) == c:
    print(c," is the correct product of multiplication")
else:
    print(c, " is incorrect product of multiplication")
    print(a*b, " is the correct product")
```

#### Q11.	Create a small DataFrame manually with 10 rows and columns: 'Name', 'Age', 'City', and 'Salary'. Then:
- Use .info(), .describe()
- Select 'Name' and 'City' columns
- Drop 'City' column
- Fill any missing values in 'Salary' column with the mean
- Remove any duplicate rows

```python
# importing the libraries and creating a datafrme 
import pandas as pd
Data = {
    "Name" : ["Atta" , "Ankit" , "Akshara" , "Bhargav" , "Bhargav" , "Dheeraj" , "Ganesh" , "Harsh" , "jay" , "Kanishka"],
    "Age" : [20,23,21,18,18,22,20,30,26,28],
    "City" : ['Pune', 'Mumbai', 'Delhi', 'Kolkata', 'Kolkata', 'Hyderabad', 'Bangalore', 'Jaipur', 'Ahmedabad', 'Lucknow'],
    "Salary" : [45000, 52000, 40000, np.nan, np.nan, 43000, 57000,np.nan , 65000, 41000]
}
df = pd.DataFrame(Data)
print(df)
```

```python
df.info() # we can observe that salary has only 7 non-null data points compared to others which have 10 
```

```python
df.describe()
```

```python
# Drop the 'City' column
df = df.drop('City', axis=1)

print(df)
```

```python
## Filling na values with mean
mean = df['Salary'].mean()
df['Salary'] = df['Salary'].fillna(mean)
print(df)
```

```python
df.duplicated() # checking for duplicated columns 
```

```python
df = df.drop_duplicates() # dropping duplicated column 
df.duplicated()
```

```python
print(df) # final cleaned data frame
```

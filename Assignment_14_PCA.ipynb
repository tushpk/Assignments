### Dimensionality Reduction using PCA 

```python
### first we standardize the data 
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
df_scaled = scaler.fit_transform(df)
```

```python
df_scaled = pd.DataFrame(df_scaled , columns = df.columns)
```

```python
df_scaled.describe()
```

### Observation

- All features have a mean approximately equal to 0, indicating successful standardization.
- The standard deviation of all features is approximately 1.
- The values of the features are now on a similar scale.
- The median values are generally close to 0 after scaling.
- The minimum and maximum values show the standardized range of each feature.
- Scaling helps prevent features with larger original values from dominating the model.


```python
###First we get how many components should we retain 
from sklearn.decomposition import PCA
pca = PCA()
pca.fit(df_scaled)

explained_variance = pca.explained_variance_ratio_
plt.figure(figsize = (8,5))
plt.plot(range(1,len(explained_variance)+1),
        explained_variance, marker = "o")
plt.xlabel('principal component')
plt.ylabel('Explained Variance Ratio')
plt.title("Scree Plot")
```

```python
cumulative_variance = pca.explained_variance_ratio_.cumsum()
plt.figure(figsize=(8, 5))
plt.plot(range(1, len(cumulative_variance) + 1),
         cumulative_variance, marker='o')

plt.xlabel('Number of Principal Components')
plt.ylabel('Cumulative Explained Variance')
plt.title('Cumulative Explained Variance')
plt.grid()
plt.show()
```

### Observation

- PC1 explains the highest variance at approximately 39.5%.
- PC2 and PC3 explain approximately 18% and 10% of the variance.
- The scree plot shows a sharp decline up to around PC3–PC4.
- After PC4, the explained variance decreases more gradually.
- The first 8 components explain approximately 92% of the total variance.
- Therefore, 8 principal components can be selected to retain about 90% of the information.

```python
### Transforming dataset into components 
pca = PCA(n_components=8)
```

```python
X = df_scaled.drop('Type', axis=1)

df_pca = pca.fit_transform(X)

df_pca = pd.DataFrame(
    df_pca,
    columns=[f'PC{i}' for i in range(1, 9)]
)
```

```python
df_pca
```

```python
### K-means
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score, davies_bouldin_score

kmeans_original = KMeans(
    n_clusters=3,
    random_state=42,
    n_init=10
)

clusters_original = kmeans_original.fit_predict(X)

df_original_clustered = X.copy()
df_original_clustered['Cluster'] = clusters_original

df_original_clustered.head()
```

```python
### evaluation
silhouette_original = silhouette_score(X, clusters_original)
db_original = davies_bouldin_score(X, clusters_original)

print("Silhouette Score:", silhouette_original)
print("Davies-Bouldin Index:", db_original)
```

```python
### visualizing clusters

pca_2d = PCA(n_components=2)
X_2d = pca_2d.fit_transform(X)

plt.figure(figsize=(8, 6))

plt.scatter(
    X_2d[:, 0],
    X_2d[:, 1],
    c=clusters_original
)

plt.xlabel('PC1')
plt.ylabel('PC2')
plt.title('K-Means Clustering - Original Data')
plt.show()
```

```python
### Application of K-means to PCA data
pca = PCA(n_components=8)

X_pca = pca.fit_transform(X)

kmeans_pca = KMeans(
    n_clusters=3,
    random_state=42,
    n_init=10
)

clusters_pca = kmeans_pca.fit_predict(X_pca)

df_pca_clustered = pd.DataFrame(
    X_pca,
    columns=[f'PC{i}' for i in range(1, 9)]
)

df_pca_clustered['Cluster'] = clusters_pca

df_pca_clustered.head()
```

```python
### Evaluation 
silhouette_pca = silhouette_score(X_pca, clusters_pca)
db_pca = davies_bouldin_score(X_pca, clusters_pca)

print("Silhouette Score:", silhouette_pca)
print("Davies-Bouldin Index:", db_pca)
```

```python
#Visualisation 
plt.figure(figsize=(8, 6))

plt.scatter(
    X_pca[:, 0],
    X_pca[:, 1],
    c=clusters_pca
)

plt.xlabel('PC1')
plt.ylabel('PC2')
plt.title('K-Means Clustering - PCA Data')
plt.show()
```

```python
### Comparision 
comparison = pd.DataFrame({
    'Dataset': ['Original Data', 'PCA Data'],
    'Silhouette Score': [silhouette_original, silhouette_pca],
    'Davies-Bouldin Index': [db_original, db_pca]
})

comparison
```

### Comparison and Analysis

- K-Means was applied to both the standardized original data and PCA-transformed data.
- The PCA-based clustering achieved a higher Silhouette Score than the original data.
- The Davies-Bouldin Index was lower for PCA data, indicating better-separated clusters.
- PCA reduced the dimensionality from 13 features to 8 components while retaining about 92% of the variance.
- The improved clustering suggests that PCA removed some redundant information and noise.
- PCA also makes the data easier to visualize and can reduce computational complexity.

### Conclusion and Insights

- K-Means clustering was successfully performed on both the original standardized data and PCA-transformed data.
- The original dataset produced reasonably separated clusters, but PCA improved the clustering performance.
- PCA reduced the number of features from 13 to 8 while retaining approximately 92% of the total variance.
- The PCA dataset achieved a higher Silhouette Score and a lower Davies-Bouldin Index.
- This shows that dimensionality reduction can remove redundant information and improve clustering quality.
- PCA is useful when datasets contain many correlated features, while direct clustering can be preferred when preserving all original features is important.

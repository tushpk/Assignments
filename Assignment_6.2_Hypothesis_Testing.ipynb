## HYPOTHESIS TESTING

### Background 
Bombay hospitality Ltd. operates a franchise model for producing exotic Norwegian dinners throughout New England. The operating cost for a franchise in a week (W) is given by the equation W = $1,000 + $5X, where X represents the number of units produced in a week. Recent feedback from restaurant owners suggests that this cost model may no longer be accurate, as their observed weekly operating costs are higher.

### Data Provided:
* The theoretical weekly operating cost model: W = $1,000 + $5X
* Sample of 25 restaurants with a mean weekly cost of Rs. 3,050
* Number of units produced in a week (X) follows a normal distribution with a mean (μ) of 600 units and a standard deviation (σ) of 25 units


```python
### Formation of Hypothesis 
# Given Weekly Cost given by W = 1000 + 5X
# Calculating the theoretical mean weekly cost  
μ = 1000 + 5*600
print(μ,"is the theoretical mean weekly cost")
```

## Null Hypothesis :
* The current cost model is accurate; the mean weekly operating cost has not increased
* H0: μ =4000
## Alternate Hypothesis :
* The mean weekly operating cost is higher than predicted by the model.
* H1: μ>4000

```python
import numpy as np
from scipy import stats
```

```python
sample_mean = 3050 #x bar
sigma = 5*25 #standard deviation
n = 25 # sample size
μ = 1000 + 5*600 # theoretical mean 

z = (sample_mean - μ)/(sigma/np.sqrt(n))
print(z,"is the test statistic(z-score)")

```

```python
## Determining the probability at 5% significance level 
alpha = 0.05 
p_value = 1 - stats.norm.cdf(z)
print("The p value is ",p_value)
if p_value<alpha:
    print("we reject the null hypothesis ")
else:
    print("We fail to reject the null hypothesis ")
```

```python
## Making a decision using z critical value 
z_critical = stats.norm.ppf(0.95)
print("z critical is",z_critical)
if z_critical<z :
    print("we reject the null hypothesis")
else:
    print("we fail to reject the null hypothesis ")
```

### Conclusion 
At the 5% significance level, there is insufficient evidence to support the restaurant owners' claim that weekly operating costs are higher than predicted by the model.

In fact, the sample mean cost: x_bar =3050

is substantially lower than the model's predicted mean cost: μ=4000

The data do not support the claim that weekly operating costs have increased. Based on the sample, the observed average operating cost is significantly lower than the cost predicted by the model.

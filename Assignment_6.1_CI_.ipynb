## ESTIMATION AND CONFIDENCE INTERVALS

### Scenario
A manufacturer of print-heads for personal computers is interested in estimating the mean durability of their print-heads in terms of the number of characters printed before failure. To assess this, the manufacturer conducts a study on a small sample of print-heads due to the destructive nature of the testing process.


### Data
A total of 15 print-heads were randomly selected and tested until failure. The durability of each print-head (in millions of characters) was recorded as follows:
1.13, 1.55, 1.43, 0.92, 1.25, 1.36, 1.32, 0.85, 1.07, 1.48, 1.20, 1.33, 1.18, 1.22, 1.29


```python
## Importing libraries
import pandas as pd 
import numpy as np 
from scipy import stats 
```

```python
## lets load the data 
data = ([1.13, 1.55, 1.43, 0.92, 1.25, 1.36, 1.32, 0.85, 1.07, 1.48, 1.20, 1.33, 1.18, 1.22, 1.29])
mean_data = np.mean(data)
std_data = np.std(data)
n = len(data)
std_err = std_data/np.sqrt(n)
print("sample mean :",mean_data)
print("sample standard diviation : ",std_data)
print(f"sample size = {n}")
print(f"standard error = {std_err}")

```

```python
## A. Building 99% Confidence Interval Using Sample Standard Deviation
## Since the population standard diviation is unknown we use the t-distribution to calculate the confidence interval 
ci = stats.t.interval(0.99 , n-1 , loc = mean_data , scale = std_err )
print(ci)
# .99 - confidence level , n-1 - degree of freedom  , loc -mean , scale standard_error 
```

### Observation 
For unkown population standard deviation We are 99% confident that the true average durability of all print-heads in the population lies between 1.0952 million and 1.3821 million characters.

```python
## B. Building 99% Confidence Interval Using Known Population Standard Deviation = 0.2
## Since the population standard diviation is now known we use the normal distribution to calculate the confidence interval
pop_std = 0.2
pop_se = 0.2/np.sqrt(n)
ci = stats.norm.interval(0.99 , loc = mean_data , scale =pop_se ) 
print(ci)
```

### Observation 
If the population standard deviation is truly known to be 0.2 million characters, we are 99% confident that the population mean durability lies between 1.1057 million and 1.3717 million characters.
